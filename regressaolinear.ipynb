```python
from sklearn.linear_model import LinearRegression


# Orçamento de Marketing
x = [[1000],[2500],[3250],[5000],[6500]]
# Lucro em Vendas
y = [10_000, 20_500, 33_250, 59_000, 77_000]


modelo = LinearRegression()


# Treinando o modelo
modelo.fit(x,y)


predicao = modelo.predict([[5000]])


print(f"Predição de lucro: R$ {predicao[0]:,.2f}")
```

```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Duração do aluguel
x = [[10], [15], [20], [25], [30], [35], [40], [45], [50], [55], [60]]

# Valores do aluguel
y = [10, 12, 14, 16, 18, 19, 20, 23, 24, 30, 32]

# Criando o modelo
modelo = LinearRegression()

# Treinando o modelo
modelo.fit(x, y)

# Previsões
y_pred = modelo.predict(x)

# Mostrando os dados no gráfico
plt.scatter(x, y, color="blue", label="Dados reais")

# Mostrando a linha de regressão
plt.plot(x, y_pred, color="red", label="Regressão linear")

# Títulos e legendas
plt.xlabel("Duração do aluguel")
plt.ylabel("Valor do aluguel")
plt.title("Regressão Linear")
plt.legend()

# Exibir gráfico
plt.show()
```
